## The scenario

Your cousin has made millions of dollars speculating on real estate. He offers to become business partners with you: he supplies the money, and you supply models that predict how much various houses are worth.

When you ask how he has predicted real estate values in the past, he says it is just intuition. More questioning reveals that he has identified price patterns in houses he has seen, and he uses those patterns to make predictions for new houses.

Machine learning works the same way. The first model in the course is the **decision tree**. Other models give more accurate predictions, but decision trees are easy to understand and they are the building block for some of the best models in data science.

## The simplest decision tree

![Sample decision tree](images/sample-decision-tree.png)

This tree divides houses into two categories. The predicted price for a house is the historical average price of the houses in the same category.

Data decides how to break the houses into two groups, and data decides the predicted price in each group. Capturing patterns from data is called **fitting** or **training** the model. The data used to fit the model is the **training data**. Once the model is fit, you apply it to new data to **predict** the prices of other homes.

## Improving the decision tree

Which of these two trees is more likely to result from fitting real estate training data?

![Two candidate trees](images/two-candidate-trees.png)

A tree with more splits captures more factors. These are called **deeper** trees. A tree that also considers the lot size of each house might look like this:

![Depth 2 tree](images/depth-2-tree.png)

You predict the price of a house by tracing through the tree, always picking the path that matches the house's characteristics. The point at the bottom where the prediction is made is called a **leaf**. The data determines the splits and the values at the leaves.
